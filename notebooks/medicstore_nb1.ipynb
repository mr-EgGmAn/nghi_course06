```python
# Download the data - done
# uv add pandas in terminal - done

# import pandas
import pandas as pd

# Load the data
df_sales_part1 = pd.read_csv("../data/Sales_MedicStore_part1.csv", sep=";")
df_sales_part2 = pd.read_csv("../data/Sales_MedicStore_part2.csv", sep=";")
df_mapping = pd.read_excel("../data/Mapping.xlsx")
```

```python
# create a list of all the sales data tables
list_sales = [df_sales_part1, df_sales_part2]

# concatenate the list of dataframes into a single dataframe
df_sales = pd.concat(list_sales)

# Examine if we lose data or not
print(f"""
      Data sales part 1: {df_sales_part1.shape}
      Data sales part 2: {df_sales_part2.shape}
      Data sales total: {df_sales.shape}
      """)
```

```python
# Look at columns of sales table and mapping table
print(f"""
      Sales columns:
      {df_sales.columns}

      Mapping cloumns:
      {df_mapping.columns}
      """)
```

```python
# Assemble the Sales Table with the Mapping Table
df_merged = pd.merge(
    df_sales, df_mapping, left_on="Sub-Category", right_on="Sub-Category"
)

# Look at the merged data
df_merged.head()
```

```python
# Reexamine if we have enough of data or not
print(f"""
      Data sales part 1: {df_sales_part1.shape}
      Data sales part 2: {df_sales_part2.shape}
      Data sales total: {df_sales.shape}
      Data sales after merged with mapping table: {df_merged.shape}
      """)
```

```python
# Create a new column Total_sales = Quantity * Unit Price
df_merged["Total_sales"] = df_merged["Quantity"] * df_merged["Unit Price"]

df_merged.head()
```

```python
# Total sales of all categories:
df_merged["Total_sales"].sum()
```

```python
# Export the resulting table in an Excel file that could be used by any other team member
df_merged.to_excel("../exported_file/final_dataset1.xlsx")
```

# OBJECT 3: Make some graph analysis


```python
# uv add pathlib, matplotlib.pyplot - done

# Import packages
from matplotlib.patches import Patch
import matplotlib.pyplot as plt
from openpyxl.drawing.image import Image
```

```python
# Make a graph of the number of visits per gender - DATASET DOESN'T HAVE GENDER INFORMATION

# Make a graph of the number of visits per department (or: Make a graph of the number of orders per Category)

# Create df count each category and sub-category' orders
df_orders_per_category = df_merged["Category"].value_counts()
df_orders_per_subcategory = df_merged["Sub-Category"].value_counts()
```

```python
fig, ax = plt.subplots()
df_orders_per_category.plot.bar(ax=ax, title="Number of orders per Category", rot=0)
fig.savefig("../exported_file/Number_of_orders_per_Category.png", bbox_inches="tight")
```

```python
# Create bar chart
categories = df_orders_per_subcategory.index.tolist()
n_bars = len(categories)

cmap = plt.get_cmap("tab20")
colors = [cmap(i % cmap.N) for i in range(n_bars)]

fig, ax = plt.subplots(figsize=(12, 6))
df_orders_per_subcategory.plot.bar(
    ax=ax,
    color=colors,
    title="Number of orders per Sub-Category",
    rot=0,
)

# Keep the color legend next to the chart
legend_handles = [
    Patch(
        facecolor=colors[i],
        label=f"{categories[i]} ({df_orders_per_subcategory.iloc[i]})",
    )
    for i in range(n_bars)
]
ax.legend(
    handles=legend_handles,
    title="Sub-Category",
    bbox_to_anchor=(1.01, 1),
    loc="upper left",
    fontsize=8,
    title_fontsize=9,
    ncol=1,
)

ax.tick_params(axis="y", labelsize=9)
ax.set_xlabel("Sub-Category", fontsize=10)
ax.set_title("Number of orders per Sub-Category", fontsize=12)
ax.set_xticklabels([])

plt.tight_layout()
fig.savefig(
    "../exported_file/Number_of_orders_per_Sub_Category.png", bbox_inches="tight"
)
```

```python
# Create a dictionary to store the analysis results


dict_technical_quest = {
    "1a. What is the number of lines and columns in Sales_MedicStore_part1.csv": df_sales_part1.shape,
    "1b. What is the number of lines and columns in Sales_MedicStore_part2.csv": df_sales_part2.shape,
    "1c. What is the number of lines and columns in Mapping.xlsx": df_mapping.shape,
    "1d. What is the number of lines and columns in final dataset": df_merged.shape,
    "2. What are the column names in the final dataset": df_merged.columns,
    "3. What are the top 5 lines of our dataset": df_merged.head(5),
    "4. What are the last 5 lines of our dataset": df_merged.tail(5),
    "5. Make a general statistical description of all the dataset": df_merged.describe(),
}


dict_business_quest = {
    "1. how many orders do we have": df_merged["Order ID"].nunique(),
    "2. what is the total sales amount": df_merged["Total_sales"].sum(),
    "3. Who is the top 3 customers that makes the most purchases (order count)": df_merged[
        "Customer Name"
    ]
    .value_counts()
    .head(3),
    "4. What is the total Sales for France": total_sales_France,
    "5. How much does the total Sales for France represent compare to the total global Sales": france_sales_per,
}


# Convert the dictionary to a DataFrame for better visualization
df_technical = pd.DataFrame(
    list(dict_technical_quest.items()), columns=["Question", "Answer"]
)
df_business = pd.DataFrame(
    list(dict_business_quest.items()), columns=["Question", "Answer"]
)
```
