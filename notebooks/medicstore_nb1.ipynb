```python
# Download the data - done
# uv add pandas in terminal - done

# import pandas
import pandas as pd

# Load the data
df_sales_part1 = pd.read_csv("../data/Sales_MedicStore_part1.csv", sep=";")
df_sales_part2 = pd.read_csv("../data/Sales_MedicStore_part2.csv", sep=";")
df_mapping = pd.read_excel("../data/Mapping.xlsx")
```

```python
# create a list of all the sales data tables
list_sales = [df_sales_part1, df_sales_part2]

# concatenate the list of dataframes into a single dataframe
df_sales = pd.concat(list_sales)

# Examine if we lose data or not
print(f"""
      Data sales part 1: {df_sales_part1.shape}
      Data sales part 2: {df_sales_part2.shape}
      Data sales total: {df_sales.shape}
      """)
```

```python
# Look at columns of sales table and mapping table
print(f"""
      Sales columns:
      {df_sales.columns}

      Mapping cloumns:
      {df_mapping.columns}
      """)
```

```python
# Assemble the Sales Table with the Mapping Table
df_merged = pd.merge(
    df_sales, df_mapping, left_on="Sub-Category", right_on="Sub-Category"
)

# Look at the merged data
df_merged.head()
```

```python
# Reexamine if we have enough of data or not
print(f"""
      Data sales part 1: {df_sales_part1.shape}
      Data sales part 2: {df_sales_part2.shape}
      Data sales total: {df_sales.shape}
      Data sales after merged with mapping table: {df_merged.shape}
      """)
```

```python
# Create a new column Total_sales = Quantity * Unit Price
df_merged["Total_sales"] = df_merged["Quantity"] * df_merged["Unit Price"]

df_merged.head()
```

```python
# Total sales of all categories:
df_merged["Total_sales"].sum()
```

```python
# Export the resulting table in an Excel file that could be used by any other team member
df_merged.to_excel("../exported_file/resulting_table.xlsx")
```
